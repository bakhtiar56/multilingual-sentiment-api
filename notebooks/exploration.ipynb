# Multilingual Sentiment – Exploratory Data Analysis

This notebook explores the `tydjohnson/multilingual-sentiments` dataset and visualises class distributions, text lengths, and language breakdowns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datasets import load_dataset

sns.set_theme(style='whitegrid')
print('Libraries loaded')
```

## 1. Load Dataset

```python
dataset = load_dataset('tydjohnson/multilingual-sentiments')
df = dataset['train'].to_pandas()
df.head()
```

## 2. Class Distribution

```python
fig, ax = plt.subplots(figsize=(6, 4))
df['label'].value_counts().plot(kind='bar', ax=ax, color=['#e74c3c', '#95a5a6', '#2ecc71'])
ax.set_title('Sentiment Class Distribution')
ax.set_xlabel('Label')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

## 3. Text Length Distribution

```python
df['text_len'] = df['text'].str.split().str.len()
fig, ax = plt.subplots(figsize=(8, 4))
sns.histplot(df['text_len'], bins=50, ax=ax)
ax.set_title('Distribution of Text Lengths (words)')
ax.set_xlabel('Number of words')
plt.tight_layout()
plt.show()

df['text_len'].describe()
```

## 4. Language Breakdown

```python
if 'language' in df.columns:
    fig, ax = plt.subplots(figsize=(10, 4))
    df['language'].value_counts().plot(kind='bar', ax=ax)
    ax.set_title('Samples per Language')
    ax.set_xlabel('Language')
    ax.set_ylabel('Count')
    plt.tight_layout()
    plt.show()
else:
    print('No language column found in dataset')
```
